```python
# הספריות מותקנות בסביבה הווירטואלית של הפרויקט.
# אם חסרות ספריות, הסר את סימן ההערה מהשורה הבאה והריץ:
# %pip install transformers torch
import sys
print(sys.executable)
```

```python
import os
from pathlib import Path

# שמירת ההורדות בתוך הפרויקט, בתיקייה שאינה מיועדת ל-Git.
os.environ["HF_HOME"] = str(Path.cwd() / ".hf-cache")
from transformers import pipeline

MODEL_ID = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"
classifier = pipeline("sentiment-analysis", model=MODEL_ID, device="cpu")
print("Model loaded:", MODEL_ID)
```

```python
reviews = [
    "I love this product. It works perfectly!",
    "The service was terrible and I want my money back.",
    "The delivery was late, but the product is excellent.",
]
results = classifier(reviews, truncation=True)
for review, result in zip(reviews, results):
    print(review)
    print(f"  {result['label']} | score = {result['score']:.3f}\n")
```

## עכשיו תורך
שנה את המשפט בתא הבא והריץ אותו שוב. נסה משפט חיובי, שלילי, ניטרלי וציני.

`score` הוא הציון שהמודל נותן לתווית שנבחרה; הוא אינו הבטחה לנכונות ואינו מדד לעוצמת הרגש. מודל יכול לטעות גם עם ציון גבוה.

נסה להשוות בין `The product is good` לבין `The product is not good` ולבדוק אם השלילה משנה את התוצאה.

```python
my_review = "The product is not good."
classifier(my_review, truncation=True)
```
